##PREPROCESSING :-

```python
df.drop(columns=["id"],inplace=True)
```

```python
df["penalty"].unique()
```

```python
df["penalty"].fillna(0,inplace=True)
```

```python
df["byes"].unique()
```

```python
df["byes"].fillna(0,inplace=True)
```

```python
print("For Noballs : ",df["noballs"].unique())
print("For Legbyee : ",df["legbyes"].unique())
print("For Wides : ",df["wides"].unique())
```

```python
df["noballs"].fillna(0,inplace=True)
df["legbyes"].fillna(0,inplace=True)
df["wides"].fillna(0,inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False).head(15)
```

```python
df.drop(columns=["other_player_dismissed","other_wicket_type","city"],inplace=True)
```

```python
df["player_dismissed"].fillna("NO One",inplace=True)
```

```python
df["wicket_type"].unique()
```

```python
df["wicket_type"].fillna("No_Wicket",inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False).head(5)
```

```python
df["umpire3"]=df["umpire3"].fillna("Unknown")
```

```python
df.columns
```

```python
df[df["winner"].isnull()][['match_id', 'season_x','team1', 'team2', 'result', 'winner']].drop_duplicates('match_id')
```

```python
df["winner"]=df["winner"].fillna("No Result")
```

```python
df[df["player_of_match"].isnull()][['player_of_match','match_id', 'season_x', 'team1', 'team2', 'result', 'winner']].drop_duplicates('match_id')
```

```python
df.loc[df["winner"]=="No Result","player_of_match"]="No Result"
```

```python
df.isnull().sum().sort_values(ascending=False).head(3)
```

```python
df["Total_run_per_ball"]=df["runs_off_bat"]+df["extras"]
```

```python
df["over"]=df["ball"].astype(int)+1
```

```python
df["Legel_Ball"]=((df["wides"]==0)&(df["noballs"]==0)).astype(int)
df['balls_bowled'] = df.groupby(['match_id', 'innings'])["Legel_Ball"].cumsum()
df["ball_left"]=120-df["balls_bowled"]
```

```python
df["current_score"]=df.groupby(['match_id','innings'])["Total_run_per_ball"].cumsum()
df["final_total"]=df.groupby(['match_id','innings'])["Total_run_per_ball"].transform('sum')
```

```python
df.columns
```

```python
df["player_dismissed"].unique()
```

```python
df['is_wicket'] = (df['player_dismissed'] != 'Not One').astype(int)
df['wickets_fallen'] = df.groupby(['match_id', 'innings'])['is_wicket'].cumsum()
```

```python
print(df['player_dismissed'].unique()[:10])
print(df['is_wicket'].mean())
print(df.groupby(['match_id', 'innings'])['wickets_fallen'].max().describe())
```
