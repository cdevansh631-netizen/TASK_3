```python
data = df[df["innings"].isin([1, 2])].copy()
print(data.shape)
```

```python
num_cols = ["innings", "over", "balls_bowled", "ball_left", "current_score", "wickets_fallen", "current_run_rate", "runs_off_bat", "extras"]
cat_cols = ["batting_team", "bowling_team", "venue_x"]

X = data[num_cols + cat_cols].copy()
y = data["final_total"]
```

```python
from sklearn.preprocessing import LabelEncoder

encoders = {}
for col in cat_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    encoders[col] = le

X.head()
```

```python
from sklearn.model_selection import train_test_split

train_matches, test_matches = train_test_split(data["match_id"].unique(), test_size=0.2, random_state=42)

train_rows = data["match_id"].isin(train_matches)
test_rows = data["match_id"].isin(test_matches)

X_train = X[train_rows]
X_test = X[test_rows]
y_train = y[train_rows]
y_test = y[test_rows]

print("Train :", X_train.shape, " Test :", X_test.shape)
```

##TO Prevent Data Leakage :-

- One Match has 240 ball and all of them have a same final score.
- A Normal Random split put data of same match in both train and test Due to which model
  already seen that data and give better result Which is Basically Fake.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])

X_train.head()
```

##PART 2 - PCA-(PRINCIPLE COMPONENT ANALYSIS)

```python
from sklearn.decomposition import PCA

pca = PCA()
pca.fit(X_train[num_cols])

evr = pca.explained_variance_ratio_
cum_evr = np.cumsum(evr)
print(evr.round(3))
print(cum_evr.round(3))
```

```python
plt.figure(figsize=(13, 5))

plt.subplot(1, 2, 1)
plt.plot(range(1, len(evr) + 1), evr, marker="o", color="blue")
plt.title("Scree Plot")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")

plt.subplot(1, 2, 2)
plt.plot(range(1, len(cum_evr) + 1), cum_evr, marker="o", color="green")
plt.axhline(0.95, color="red", linestyle="--")
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Variance")

plt.tight_layout()
plt.show()
```

```python
n_comp = np.argmax(cum_evr >= 0.95) + 1
print("Components needed for 95% variance :", n_comp, "out of", len(num_cols))

pca_final = PCA(n_components=n_comp)
X_train_pca = pca_final.fit_transform(X_train[num_cols])
X_test_pca = pca_final.transform(X_test[num_cols])
print(X_train_pca.shape)
```

- A Principle Component is a new column which is created by mixing numerious old columns .Where PC1 Gives the direction where data is spread out the most Therefore hold most information And PC2 is next best direction.In my data it is showing how far is inning going. 

- PCA help in dimensionality reduction As it tranform higer dimension data into lower dimension .Here Pca join such repeated columns who saying the same thing .Help to reduce column column and give help faster training.

#PART 3-(UNSUPERVISED LEARNING):-

##K-Means:-
- It Split the data into k group(cluster) which is give by us And placed random k centroid in the data.
  And After that it repeat continous 2 step:-
  Assignment(Every pt goes to nearest centroid) -> Update:every centroid updates to the mean of pts assigned.These step continue untill centroid stop moving from its position.e
###To Calculate K:-
- We have 2 way Elbow Method(wcss V/S no of k(ie-1,2,3,4,5 etc)),silhoutee Score(by Measuring silhoutee score of pt from other pts by b-a/max(b,a) Which give Highest score eill choosen as k).

###Problem Of K-means:-
- it work only with data have round shape doesn't work well on iregular shape or elongated shape.
- It is very sensitive to outlier(As Euclid formula involve).
- It doesn't work well for data have Different densities.


##DBSCAN:-(Density Based Spatial Clustering of Application With noise):-

- This Approack is density based approach and work well on iregualar shape/elongated shapes.
  As Their is no distance based formula involve so work well on outlier.

  It has 2 main hyperperimeter :
  1->Epsilin :- It is a Radius under Which if a pt lie it belong to  that cluster.
  2->Minpts :- They ar minimum pts within neighbouhood radius required for a area to be dense.




#PART-4-(Essemble_Teachnique):  

- I choose a regression Problem Type In Which My Target is to predict total score of a inning .


```python
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV
```

```python
rf = RandomForestRegressor(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
```

```python
ada = AdaBoostRegressor(n_estimators=50, random_state=42)
ada.fit(X_train, y_train)
ada_pred = ada.predict(X_test)
```

```python
xgb = XGBRegressor(n_estimators=200, learning_rate=0.1, max_depth=6, random_state=42)
xgb.fit(X_train, y_train)
xgb_pred = xgb.predict(X_test)
```

```python
def get_scores(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return {"Model": name, "MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2}

results = pd.DataFrame([
    get_scores("Random Forest", y_test, rf_pred),
    get_scores("AdaBoost", y_test, ada_pred),
    get_scores("XGBoost", y_test, xgb_pred),
])
results.round(3)
```

```python
best = results.sort_values("R2", ascending=False).iloc[0]
print("Best model :", best["Model"], "with R2 =", round(best["R2"], 3), "and MAE =", round(best["MAE"], 2))
```
